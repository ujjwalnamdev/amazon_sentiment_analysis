```python
import numpy as np
import pandas as pd
import torch
import cv2
import datasets

print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("PyTorch:", torch.__version__)
print("OpenCV:", cv2.__version__)
print("Datasets:", datasets.__version__)
print("CUDA:", torch.cuda.is_available())
```

```python
from datasets import load_dataset

digital_music_reviews = load_dataset(
    "json",
    data_files="dataset/Digital_Music.jsonl",
    split="train",
    streaming=True
)
```

```python
#SPLIT THE DATSET INTO 20000 ROWS
sample1 = list(digital_music_reviews.take(20000))
```

```python
df1 = pd.DataFrame(sample1)
print(df1.shape)
print(df1.columns)
df1.head()
```

```python
#2.load Arts_Crafts_and_Sewing_dataset
Arts_crafts_and_sewings = load_dataset(
    "json",
    data_files = "dataset/Arts_Crafts_and_Sewing.jsonl",
    split = "train",
    streaming = True
)
```

```python
sample2 = list(Arts_crafts_and_sewings.take(20000))
```

```python
df2 = pd.DataFrame(sample2)
print(df2.shape)
print(df2.columns)
df2.head()
```

```python
#3.load electronics dataset
Electronics_reviews = load_dataset(
    "json",
    data_files = "dataset/Electronics.jsonl",
    split = "train",
    streaming = True
)
sample3 = list(Electronics_reviews.take(20000))
```

```python
df3 = pd.DataFrame(sample3)
print(df3.shape)
print(df3.columns)
df3.head()
```

```python
#handmade Products dataset
handmade_products_reviews = load_dataset(
    "json",
    data_files="dataset/Handmade_Products.jsonl",
    split="train",
    streaming=True
)
sample4 = list(handmade_products_reviews.take(20000))
```

```python
df4 = pd.DataFrame(sample4)
print(df4.shape)
print(df4.columns)
df4.head()
```

```python
health_and_personal_care_reviews = load_dataset(
    "json",
    data_files="dataset/Health_and_Personal_Care.jsonl",
    split="train",
    streaming=True
)

sample5 = list(health_and_personal_care_reviews.take(20000))
```

```python
df5 = pd.DataFrame(sample5)
print(df5.shape)
print(df5.columns)
df5.head()
```

```python
#Add categories to all dataframes
df1["category"] = 'Digital_Music'
df2["category"] = 'Arts_Crafts_and_Sewing'
df3['category'] = 'Electronics'
df4['category'] = 'Handmade_Products'
df5['category'] = 'Health_and_Personal_Care'
```

```python
for df in [df1, df2, df3, df4, df5]:
    print(df["category"].value_counts())
```

```python
import pandas as pd

universal_df = pd.concat(
    [df1, df2, df3, df4, df5],
    ignore_index = True
)
```

```python
print("Shape:", universal_df.shape)
print("\nColumns:", universal_df.columns.tolist())

print("\nFirst 5 records:")
display(universal_df.head())
```

```python
#Save universal database as a parquet file

universal_df.to_parquet(
    "universal_dataset.parquet",
    index = False
)
```
